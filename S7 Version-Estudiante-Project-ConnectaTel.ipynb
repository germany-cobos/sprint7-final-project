### 3.1 Corregir sentinels y fechas imposibles
**🎯 Objetivo:**  
Aplicar reglas de limpieza para reemplazar valores sentinels y corregir fechas imposibles.

**Instrucciones:**  
- En `age`, reemplaza el sentinel **-999** con la mediana.
- En `city`, reemplaza el sentinel `"?"` por valores nulos (`pd.NA`).  
- Marca como nulas (`pd.NA`) las fechas fuera de rango.

```python
# Reemplazar -999 por la mediana de age
age_mediana = users.loc[users['age'] != -999, 'age'].median()
users['age'] = users['age'].replace(-999, age_mediana)

# Verificar cambios
print(users['age'].describe())
```

```python

# Reemplazar ? por NA en city
users['city'] = users['city'].replace('?', pd.NA)

# Verificar cambios
print(users['city'].value_counts(dropna=False))
```

```python
# Marcar fechas futuras como NA para reg_date
users.loc[users['reg_date'].dt.year > 2024, 'reg_date'] = pd.NaT

# Verificar cambios
print(users['reg_date'].dt.year.value_counts(dropna=False))
```

## 🧩Paso 4: Summary statistics de uso por usuario


### 4.1 Agrupación por comportamiento de uso

🎯**Objetivo**: Resumir las variables clave de la tabla `usage` **por usuario**, creando métricas que representen su comportamiento real de uso histórico. 

**Instrucciones:**: 
1. Construye una tabla agregada de `usage` por `user_id` que incluya:
- número total de mensajes  
- número total de llamadas  
- total de minutos de llamadas

2. Renombra las columnas para que tengan nombres claros:  
- `cant_mensajes`  
- `cant_llamadas`  
- `cant_minutos_llamada`
3. Combina esta tabla con `users`.

```python

# Columnas auxiliares
usage["is_text"] = (usage["type"] == "text").astype(int)  # total de mensajes
usage["is_call"] = (usage["type"] == "call").astype(int)  # total de llamadas

# Agrupar información por usuario
usage_agg = usage.groupby("user_id").agg({
    "is_text": "sum",
    "is_call": "sum",
    "duration": "sum"
}).reset_index()

# observar resultado
usage_agg.head(3)
```

```python


# Renombrar columnas
usage_agg = usage_agg.rename(columns={
    "is_text": "cant_mensajes",
    "is_call": "cant_llamadas",
    "duration": "cant_minutos_llamada"
})

# observar resultado
usage_agg.head(3)
```

```python
# Combinar la tabla agregada con el dataset de usuarios
user_profile = users.merge(usage_agg, on="user_id", how="left")

# observar resultado final
user_profile.head(5)
```

💡Insights: 
Age:
La variable age no presenta outliers significativos. La mayoría de las edades se concentran en rangos normales y coherentes para clientes de telecomunicaciones.

cant_mensajes:

cant_llamadas:
La variable cant_llamadas también muestra outliers superiores, lo que sugiere la existencia de usuarios con comportamiento intensivo en llamadas.

cant_minutos_llamada:
La variable cant_minutos_llamada presenta varios outliers altos y una fuerte asimetría positiva, reflejando usuarios con consumos muy elevados de minutos.

```python

# Revisión de límites usando IQR
columnas_limites = ['cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']

for col in columnas_limites:
    Q1 = user_profile[col].quantile(0.25)
    Q3 = user_profile[col].quantile(0.75)
    IQR = Q3 - Q1
    
    limite_superior = Q3 + 1.5 * IQR
    
    print(f"\nVariable: {col}")
    print("Límite superior:", limite_superior)
    print("Valor máximo:", user_profile[col].max())
```

```python


# Revisa los limites superiores y el max, para tomar la decisión de mantener los outliers o no
user_profile[columnas_limites].describe()
```

💡Insights: 
cant_mensajes:
Se recomienda mantener los outliers, ya que representan usuarios con un comportamiento real de uso intensivo de mensajes. Eliminarlos podría ocultar patrones importantes de consumo y afectar el análisis de segmentación de clientes.

cant_llamadas:
Se recomienda mantener los outliers porque corresponden a clientes que realizan una cantidad inusualmente alta de llamadas, lo cual puede ser relevante para identificar usuarios de alto consumo o posibles clientes premium.

cant_minutos_llamada:
Se recomienda mantener los outliers, ya que reflejan usuarios con consumo elevado de minutos y forman parte del comportamiento natural del negocio. Estos valores pueden aportar información valiosa para estrategias comerciales y análisis de planes.

## 🧩Paso 6: Segmentación de Clientes

### 6.1 Segmentación de Clientes Por Uso

🎯 **Objetivo:** Clasificar a cada usuario en un grupo de uso (Bajo uso, Uso medio, Alto uso) basándose en la cantidad de llamadas y mensajes registrados.

**Instrucciones:**  
- Crea una nueva columna llamada `grupo_uso` en el dataframe `user_profile`.
- Usa comparaciones lógicas (<, >) para evaluar las condiciones de llamadas y mensajes y asigna:
  - `'Bajo uso'` cuando llamadas < 5 y mensajes < 5
  - `'Uso medio'` cuando llamadas < 10 y mensajes < 10
  - `'Alto uso'` para el resto de casos

```python


# Crear columna grupo_uso

user_profile.loc[
    (user_profile['cant_llamadas'] < 5) &
    (user_profile['cant_mensajes'] < 5),
    'grupo_uso'
] = 'Bajo uso'

user_profile.loc[
    (user_profile['cant_llamadas'] < 10) &
    (user_profile['cant_mensajes'] < 10) &
    (user_profile['grupo_uso'].isna()),
    'grupo_uso'
] = 'Uso medio'

user_profile.loc[
    user_profile['grupo_uso'].isna(),
    'grupo_uso'
] = 'Alto uso'

```

```python
# Crear columna grupo_uso

user_profile.loc[
    (user_profile['cant_llamadas'] < 5) & 
    (user_profile['cant_mensajes'] < 5),
    'grupo_edad'
] = 'Bajo uso'

user_profile.loc[
    (user_profile['cant_llamadas'] < 10) & 
    (user_profile['cant_mensajes'] < 10) &
    (user_profile['grupo_edad'].isna()),
    'grupo_edad'
] = 'Uso medio'

user_profile.loc[
    user_profile['grupo_edad'].isna(),
    'grupo_edad'
] = 'Alto uso'
```
